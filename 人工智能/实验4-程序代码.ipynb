```python
import numpy as np
import operator
import os

def classify0(inX, dataSet, labels, k):
    """
    kNN分类算法
    参数:
        inX - 待分类的数据点
        dataSet - 训练数据集
        labels - 训练数据标签  
        k - 近邻数
    返回:
        预测的类别
    """
    # 计算距离 - 步骤1
    dataSetSize = dataSet.shape[0]
    
    # 使用tile函数将inX复制成与dataSet相同行数的矩阵，然后相减
    diffMat = np.tile(inX, (dataSetSize, 1)) - dataSet
    
    # 计算差的平方
    sqDiffMat = diffMat ** 2
    
    # 按行求和，得到平方距离
    sqDistances = sqDiffMat.sum(axis=1)
    
    # 开方得到欧式距离
    distances = sqDistances ** 0.5
    
    # 距离排序 - 步骤2
    # argsort返回的是从小到大排序后的索引值
    sortedDistIndicies = distances.argsort()
    
    # 统计前k个最近邻的类别 - 步骤3和4
    classCount = {}
    for i in range(k):
        # 获取第i个最近邻的标签
        voteIlabel = labels[sortedDistIndicies[i]]
        # 统计该标签出现的次数
        classCount[voteIlabel] = classCount.get(voteIlabel, 0) + 1
    
    # 对统计结果进行排序 - 步骤5
    # 按照出现次数从高到低排序
    sortedClassCount = sorted(classCount.items(), 
                             key=operator.itemgetter(1), reverse=True)
    
    # 返回出现次数最多的类别
    return sortedClassCount[0][0]

# 创建测试数据集
def createDataSet():
    """
    创建测试数据集
    返回:
        group - 特征数据集
        labels - 分类标签
    """
    # 6个样本，每个样本2个特征（理综成绩，文综成绩）
    group = np.array([
        [250, 100],  # 样本1
        [270, 120],  # 样本2  
        [111, 230],  # 样本3
        [130, 260],  # 样本4
        [200, 80],   # 样本5
        [70, 190]    # 样本6
    ])
    
    # 对应的分类标签
    labels = ['理科生', '理科生', '文科生', '文科生', '理科生', '文科生']
    
    return group, labels

# 测试kNN算法
if __name__ == '__main__':
    # 创建数据集
    group, labels = createDataSet()
    
    # 待分类的测试点 (理综成绩, 文综成绩)
    test_point = [105, 210]
    
    # 使用kNN算法进行分类预测，k=3
    result = classify0(test_point, group, labels, 3)
    
    # 输出结果
    print("训练数据集：")
    for i in range(len(group)):
        print(f"样本{i+1}: 理综{group[i][0]}, 文综{group[i][1]} -> {labels[i]}")
    
    print(f"\n待分类数据: 理综{test_point[0]}, 文综{test_point[1]}")
    print(f"kNN分类结果(k=3): {result}")
```
